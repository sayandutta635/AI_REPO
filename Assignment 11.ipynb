```python
import numpy as np
import matplotlib.pyplot as plt

# Input values
x = np.linspace(-10, 10, 400)

# Activation functions
binary_step = np.where(x >= 0, 1, 0)
linear = x
sigmoid = 1 / (1 + np.exp(-x))
tanh = np.tanh(x)
relu = np.maximum(0, x)
leaky_relu = np.where(x >= 0, x, 0.01 * x)

# Plot activation functions
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.plot(x, binary_step)
plt.title("Binary Step")
plt.grid()

plt.subplot(2, 3, 2)
plt.plot(x, linear)
plt.title("Linear")
plt.grid()

plt.subplot(2, 3, 3)
plt.plot(x, sigmoid)
plt.title("Sigmoid")
plt.grid()

plt.subplot(2, 3, 4)
plt.plot(x, tanh)
plt.title("Tanh")
plt.grid()

plt.subplot(2, 3, 5)
plt.plot(x, relu)
plt.title("ReLU")
plt.grid()

plt.subplot(2, 3, 6)
plt.plot(x, leaky_relu)
plt.title("Leaky ReLU")
plt.grid()

plt.tight_layout()
plt.show()
```
