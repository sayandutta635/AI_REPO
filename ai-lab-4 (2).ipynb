# BFS

```python
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    traversal = []

    visited.add(start)

    while queue:
        node = queue.popleft()
        traversal.append(node)

        for neighbour in graph[node]:
            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)
    return traversal
result = bfs(graph, 'A')
print("BFS Traversal:")
print(" -> ".join(result))
```

```python
from collections import deque
import networkx as nx
import matplotlib.pyplot as plt


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    traversal = []

    visited.add(start)

    while queue:
        node = queue.popleft()
        traversal.append(node)

        for neighbour in graph[node]:
            if neighbour not in visited:
                visited.add(neighbour)
                queue.append(neighbour)

    return traversal

result = bfs(graph, 'A')
print("BFS Traversal:")
print(" -> ".join(result))

G = nx.DiGraph()

for node in graph:
    for neighbour in graph[node]:
        G.add_edge(node, neighbour)
pos = {
	  'A': (0, 3),
    'B': (-1, 2),
    'C': (1, 2),
    'D': (-1.5, 1),
    'E': (-0.5, 1),
    'F': (1, 1),
    'G': (-0.5, 0)
}
plt.figure(figsize=(7, 5))
nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=2000,
    node_color='Limegreen',
    font_size=12,
    font_weight='bold',
    arrows=True,
    arrowsize=20,
    edge_color='black',
    width=2
)

plt.title("Graph Representation for BFS")
plt.show()
```

# DFS

```python
import networkx as nx
import matplotlib.pyplot as plt

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

def dfs(graph, node, visited=None, traversal=None):
    if visited is None:
        visited = set()
    if traversal is None:
        traversal = []

    visited.add(node)
    traversal.append(node)

    for neighbour in graph[node]:
        if neighbour not in visited:
            dfs(graph, neighbour, visited, traversal)

    return traversal

result = dfs(graph, 'A')

print("DFS Traversal:")
print(" -> ".join(result))

G = nx.DiGraph()

for node in graph:
    for neighbour in graph[node]:
        G.add_edge(node, neighbour)

pos = {
    'A': (0, 3),
    'B': (-1, 2),
    'C': (1, 2),
    'D': (-1.5, 1),
    'E': (-0.5, 1),
    'F': (1, 1),
    'G': (-0.5, 0)
}

plt.figure(figsize=(7, 5))
nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=2500,
    node_color="lightgreen",
    font_size=12,
    font_weight="bold",
    arrows=True,
    arrowsize=20,
    width=2,
    edge_color="black"
)

plt.title("Graph Representation for DFS")
plt.show()
```
