# Develop a Python program to implement a rule-based Expert system for student performance evaluation. The system should take inputs such as attendance, marks, assignment performance, and internal assessment and use IF-THEN rules to classify the student as Excellent, good, average, poor.

```python
def evaluate_student(attendance, marks, assignment, internal):
    """
    Evaluate student performance using IF-THEN rules.
    
    Inputs:
        attendance : Attendance percentage (0-100)
        marks      : Exam marks percentage (0-100)
        assignment : Assignment performance percentage (0-100)
        internal   : Internal assessment percentage (0-100)
    
    Returns:
        Performance classification
    """

    if (attendance >= 90 and marks >= 85 and
            assignment >= 85 and internal >= 85):
        return "Excellent"

    elif (attendance >= 80 and marks >= 70 and
          assignment >= 70 and internal >= 70):
        return "Good"

    elif (attendance >= 70 and marks >= 50 and
          assignment >= 50 and internal >= 50):
        return "Average"

    else:
        return "Poor"


print("==========================================")
print(" Student Performance Expert System")
print("==========================================")

try:
    attendance = float(input("Enter attendance percentage: "))
    marks = float(input("Enter examination marks percentage: "))
    assignment = float(input("Enter assignment performance percentage: "))
    internal = float(input("Enter internal assessment percentage: "))

    if not all(0 <= value <= 100 for value in
               [attendance, marks, assignment, internal]):
        print("Error: All values must be between 0 and 100.")
    else:
        result = evaluate_student(
            attendance, marks, assignment, internal
        )

        print("\n--- Evaluation Result ---")
        print(f"Attendance          : {attendance}%")
        print(f"Exam Marks          : {marks}%")
        print(f"Assignment          : {assignment}%")
        print(f"Internal Assessment : {internal}%")
        print(f"Performance         : {result}")

except ValueError:
    print("Error: Please enter valid numeric values.")
```
