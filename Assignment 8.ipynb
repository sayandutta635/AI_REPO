# WAP IN PYTHON TO IMPLEMENT THE MINIMAX ALGORITHM FOR A TWO PLAYER GAME AND SHOW THE PATH
1. IMPLEMENT THE MINIMAX  ALGORITHM USING RECURSION.
2. CONSIDER MAX AND MIN PLAYERS.
3. USE THE FOLLOWING LEAF-NODE VALUES:
	3, 5, 2, 9, 12, 5, 23, 23.
4. DETERMINE THE OPTIMAL SCORE FOR THE MAX PLAYER.
5. DISPLAY THE FINAL OPTIMAL SCORE AS OUTPUT.

```python
# WAP in Python to implement Minimax Algorithm
# Two Player Game

# Leaf node values
leaf_values = [3, 5, 2, 9, 12, 5, 23, 23]


# ------------------------------------------------
# Minimax Algorithm using Recursion
# ------------------------------------------------

def minimax(depth, node_index, is_maximizing):

    # Base condition:
    # If we reach a leaf node, return its value
    if depth == 3:
        return leaf_values[node_index]

    # MAX player
    if is_maximizing:

        left = minimax(
            depth + 1,
            node_index * 2,
            False
        )

        right = minimax(
            depth + 1,
            node_index * 2 + 1,
            False
        )

        return max(left, right)

    # MIN player
    else:

        left = minimax(
            depth + 1,
            node_index * 2,
            True
        )

        right = minimax(
            depth + 1,
            node_index * 2 + 1,
            True
        )

        return min(left, right)


# ------------------------------------------------
# Main Program
# ------------------------------------------------

print("MINIMAX ALGORITHM")
print("-----------------")

print("Leaf Node Values:")
print(leaf_values)

# MAX starts at depth 0 and node index 0
optimal_score = minimax(0, 0, True)

print("\nOptimal Score for MAX Player:", optimal_score)
```

# WAP IN PY TO IMPLEMENT THE ALPHA BETA PRUNING ALGO FOR A TWO PLAYER GAME. USE THE LEAF-NODE VALUES 3 5 2 9 12 5 23 23. DETERMINE THE OPTICAL SCORE FOR THE MAX PLAYER AND DISPLAY WHICH BRANCHES ARE PRUNED DURING THE SEARCH

```python
# Alpha-Beta Pruning Algorithm
# Leaf nodes: 3, 5, 2, 9, 12, 5, 23, 23

def alpha_beta(depth, node_index, maximizing_player, values, alpha, beta):
    # Leaf node
    if depth == 3:
        return values[node_index]

    if maximizing_player:
        best_value = float('-inf')

        for i in range(2):
            value = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                False,
                values,
                alpha,
                beta
            )

            best_value = max(best_value, value)
            alpha = max(alpha, best_value)

            # Beta cutoff
            if alpha >= beta:
                print("Pruned branch at node:", node_index * 2 + i)
                break

        return best_value

    else:
        best_value = float('inf')

        for i in range(2):
            value = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                True,
                values,
                alpha,
                beta
            )

            best_value = min(best_value, value)
            beta = min(beta, best_value)

            # Alpha cutoff
            if alpha >= beta:
                print("Pruned branch at node:", node_index * 2 + i)
                break

        return best_value


# Leaf-node values
values = [3, 5, 2, 9, 12, 5, 23, 23]

# Initial alpha and beta values
alpha = float('-inf')
beta = float('inf')

# Start Alpha-Beta Pruning
optimal_score = alpha_beta(0, 0, True, values, alpha, beta)

print("\nLeaf Node Values:", values)
print("Optimal Score for MAX Player:", optimal_score)
```
